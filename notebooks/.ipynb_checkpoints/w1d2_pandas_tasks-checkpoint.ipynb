# Performing filter, groupby, merge, pivot_table.

```python
import pandas as pd
```

```python
# Filter states/UTs with literacy rates of 80% or higher
df = pd.read_csv('india_state_2011.csv')
```

```python
df
```

```python
df1 = df[df['Literacy_Rate_2011'] >= 80 ]
```

```python
df1
```

```python
#GroupBy using Region
group_by_region = df.groupby("Region").count()
```

```python
group_by_region
```

```python
#groupby using region and population using sum() function 
df.groupby("Region")["Population_2011"].sum()
```

```python
#groupby using region and population using mean() function 
df.groupby("Region")["Population_2011"].mean()
```

```python
#groupby using region and population using min() function 
df.groupby("Region")["Population_2011"].min()
```

```python
#groupby using region and population using max() function 
df.groupby("Region")["Population_2011"].max()
```

```python
#Merging Population and Literacy Rate on State using merge function
```

```python
pop_df = df[["State_UT", "Population_2011"]]
```

```python
pop_df
```

```python
lit_df = df[["State_UT", "Literacy_Rate_2011"]]
```

```python
lit_df
```

```python
merged_df = pd.merge(pop_df, lit_df, on = "State_UT")
```

```python
merged_df
```

```python
#Pivot_Table using populationa and calculating the mean using agg function
pivot = pd.pivot_table(
    df,
    index="Region",
    values="Population_2011",
    aggfunc="mean"
)

print(pivot)
```

# Export cleaned DataFrame to CSV and Parquet. Compare file sizes.

```python
import os
```

```python
# Remove rows with missing values
cleaned_df = df.dropna()
```

```python
# Save cleaned files
cleaned_df.to_csv("india_state_2011_cleaned.csv", index=False)
cleaned_df.to_parquet("india_state_2011_cleaned.parquet", index=False)
```

```python
# Compare file sizes
csv_size = os.path.getsize("india_state_2011_cleaned.csv")
parquet_size = os.path.getsize("india_state_2011_cleaned.parquet")
```

```python
print("CSV size:", round(csv_size / 1024, 2), "KB")
print("Parquet size:", round(parquet_size / 1024, 2), "KB")
```
