# Task 2: Broadcasting, Vectorised Operations, and Matrix Multiplication

```python
#Broadcasting
salary = np.array([[45, 67, 99], [69, 34, 89]])
```

```python
bonus = np.array([6, 9, 4])
```

```python
new_salary = salary + bonus
```

```python
print(new_salary)
```

```python
#also using scalar representation
new_sal = salary + 6
print(new_sal)
```

```python
#Vectorised Operations
a = np.array([2,45,67,34,78,98])

square = a ** 2
print(square)
```

```python
double = a * 2
add = a + 2
sub = a - 1
div = a/2
```

```python
print(double)
print(add)
print(sub)
print(div)
```

```python
#Matrix Multiplication
m = np.array([
    [1,3],
    [1,5]
])

n = np.array([
    [2,4],
    [4,5]
])

q = np.full([2,3], 2)
result = m * n
print(result)
print(q)
```

```python
res = m @ n
print(res)

np.matmul(m,q)
```

```python
#Mean of the array
a1 = np.array([12,4,45,5,462,12,4,35])
res1 = np.mean(a1)
print(res1)
```

```python
#Standard Deviation of the Array
res2 = np.std(a1)
print(res2)
```

```python
#Correlation on real csv dataset
file_path = Path("../data/data.csv")
data = np.genfromtxt(file_path, delimiter = ',')
print(data)
```

```python
cor = np.corrcoef(data, rowvar=False)
cor = cor.astype('int32')
print(cor)
```
