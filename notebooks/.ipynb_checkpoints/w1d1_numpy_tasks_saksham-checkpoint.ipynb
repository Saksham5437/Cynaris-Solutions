# Task 2: Broadcasting, Vectorised Operations, and Matrix Multiplication

```python
#Broadcasting
salary = np.array([[45, 67, 99], [69, 34, 89]])
```

```python
bonus = np.array([6, 9, 4])
```

```python
new_salary = salary + bonus
```

```python
print(new_salary)
```

```python
#also using scalar representation
new_sal = salary + 6
print(new_sal)
```

```python
#Vectorised Operations
a = np.array([2,45,67,34,78,98])

square = a ** 2
print(square)
```

```python
double = a * 2
add = a + 2
sub = a - 1
div = a/2
```

```python
print(double)
print(add)
print(sub)
print(div)
```

```python
#Matrix Multiplication
m = np.array([
    [1,3],
    [1,5]
])

n = np.array([
    [2,4],
    [4,5]
])

result = m * n
print(result)
```

```python
res = m @ n
print(res)
```
